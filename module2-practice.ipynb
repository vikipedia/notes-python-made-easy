```python
for f in files:
    print(f)
```

```python
txtfiles = []
for f in files:
    if find_extension(f) == "txt":
        txtfiles.append(f)
```

```python
txtfiles
```

```python
txtfiles = []
pdffiles = []
htmlfiles = []
for f in files:
    if f.endswith(".txt"):
        txtfiles.append(f)
    elif f.endswith(".pdf"):
        pdffiles.append(f)
    elif f.endswith(".html"):
        htmlfiles.append(f)
```
