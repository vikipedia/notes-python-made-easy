````python
def sumofsquares(x, y):
    return square(x) + square(y)
````

````python
def cube(x):
    return x**3
    
def sumofcubes(x, y):
    return cube(x) + cube(y)
````

````python
def sumof(x, y, func):
    return func(x) + func(y)
````
